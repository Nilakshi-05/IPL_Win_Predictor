```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
delivery = pd.read_csv("../data/raw/deliveries.csv")
match = pd.read_csv("../data/raw/matches.csv")
```

```python
total_score_df = delivery.groupby(["match_id","inning"]).sum()["total_runs"].reset_index()
```

```python
total_score_df = total_score_df[total_score_df["inning"] == 1]
```

```python
total_score_df['total_runs'] = total_score_df['total_runs'] + 1
```

```python
total_score_df
```

```python
match_df = match.merge(total_score_df[["match_id","total_runs"]],left_on = "id", right_on = "match_id")
```

```python
match_df
```

```python
match_df["team1"].unique()
```

```python
 teams = [
     "Sunrisers Hyderabad",
     'Mumbai Indians',
     'Royal Challengers Bangalore', 
     'Kolkata Knight Riders',
     'Kings XI Punjab',
     'Chennai Super Kings',
     'Rajasthan Royals',
     'Delhi Capitals',
     "Gujarat Titans",
      "Lucknow Super Giants"
 ]
```

```python
match_df["team1"] = match_df["team1"].str.replace("Delhi Daredevils", "Delhi Capitals")
match_df["team2"] = match_df["team2"].str.replace("Delhi Daredevils", "Delhi Capitals")
match_df["team1"] = match_df["team1"].str.replace('Gujarat Lions', "Gujarat Titans")
match_df["team2"] = match_df["team2"].str.replace('Gujarat Lions', "Gujarat Titans")
match_df["team1"] = match_df["team1"].str.replace('Rising Pune Supergiants', "Lucknow Super Giants")
match_df["team2"] = match_df["team2"].str.replace('Rising Pune Supergiants', "Lucknow Super Giants")
match_df["team1"] = match_df["team1"].str.replace('Deccan Chargers',  'Sunrisers Hyderabad')
match_df["team2"] = match_df["team2"].str.replace('Deccan Chargers',  'Sunrisers Hyderabad')
```

```python
match_df = match_df[match_df["team1"].isin(teams)]
match_df = match_df[match_df["team2"].isin(teams)]
```

```python
match_df
```

```python
match_df = match_df[match_df["dl_applied"] == 0]
```

```python
match_df = match_df[['match_id','city','winner','total_runs']]
```

```python
delivery_df = match_df.merge(delivery,on = 'match_id')
```

```python
delivery_df = delivery_df[delivery_df['inning'] == 2]
```

```python
delivery_df.head(50)
```

```python
 delivery_df["current_score"] =delivery_df.groupby('match_id')['total_runs_y'].cumsum()
```

```python
delivery_df["runs_left"] = delivery_df["total_runs_x"] - delivery_df['current_score']
```

```python
delivery_df["ball_left"] = 120 - ((delivery_df['over'] - 1) * 6 + delivery_df['ball'])
```

```python
 delivery_df
```

```python
delivery_df['is_wicket'] = delivery_df['player_dismissed'].notnull().astype(int)
delivery_df['wickets_fallen'] = delivery_df.groupby('match_id')['is_wicket'].cumsum()
delivery_df['wickets_left'] = 10 - delivery_df['wickets_fallen']
delivery_df[['player_dismissed', 'is_wicket', 'wickets_fallen', 'wickets_left']].tail(5)
```

```python
delivery_df["crr"] = (delivery_df["current_score" ] * 6)/(120 - delivery_df["ball_left"])
```

```python
delivery_df['rrr'] = (delivery_df['runs_left'] * 6)/ delivery_df["ball_left"]
```

```python
delivery_df
```

```python
def result(row):
    return 1 if row["batting_team"] == row['winner'] else 0
```

```python
delivery_df['result'] = delivery_df.apply(result, axis = 1)
```

```python
final_df = delivery_df[["batting_team", "bowling_team", "city", "runs_left", "ball_left", "wickets_left", "total_runs_x", "crr", "rrr", "result"]]
```

```python
final_df.sample(final_df.shape[0])
```

```python
team_map = {
    'Delhi Daredevils': 'Delhi Capitals',
    'Deccan Chargers': 'Sunrisers Hyderabad',
    "Gujarat Lions" : "Gujarat Titans",
    "Rising Pune Supergiants" : "Lucknow Super Giants"
    
}
for col in ['batting_team', 'bowling_team']:
    final_df[col] = final_df[col].replace(team_map)
```
